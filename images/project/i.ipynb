```python
from PIL import Image

def resize_image(input_path, output_path, target_size=(920, 643)):
    # Open the image
    with Image.open(input_path) as img:
        # Resize the image using LANCZOS resampling for better quality
        resized_img = img.resize(target_size, Image.Resampling.LANCZOS)
        # Save the resized image
        resized_img.save(output_path)

# Example usage
input_path = "uvu_results.png"  # Replace with your image path
output_path = "uvu_results_resized.png"  # Replace with desired output path
resize_image(input_path, output_path)
```
